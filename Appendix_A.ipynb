```python
import torch.nn.functional as F
from torch.autograd import grad
```

```python
y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.2], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)
```

```python
z = x1*w1 + b
a = torch.sigmoid(z)

loss = F.binary_cross_entropy(a,y)

grad_L_w1 = grad(loss, w1, retain_graph=True)
grad_L_b = grad(loss, b, retain_graph=True)

# by default pytorch destroys the computational graph after calculating the gradient to free the memory. however, since we will reuse this graph, we set retain_graph=True so that it stays in memory.
```

```python
print(grad_L_b)
print(grad_L_w1)
```

```python
## below code implements a classic multilayer perceptron with two hidden layers.
```

```python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(

        #1st Hdden layer
        torch.nn.Linear(num_inputs,30),
        torch.nn.ReLU(),

        #2nd Hidden layer
        torch.nn.Linear(30,20),
        torch.nn.ReLU(),

        #Output layer
        torch.nn.Linear(20,num_outputs),

        )

    def forward(self,x):
        logits = self.layers(x)
        return logits
```

```python
# Coding the number of inputs and output as variable allows us to reuse the same code for datasets with different number of features and classes.
# outputs of last layer are called logits
```

```python
model = NeuralNetwork(50,3)
```

## Setting up efficient data loaders

```python
X_train = torch.tensor([
    [-1.2,3.1],
    [-0.9,2.9],
    [-0.5,2.6],
    [2.3,-1.1],
    [2.7,-1.5]
])

y_train = torch.tensor([0,0,0,1,1])

X_test = torch.tensor([
    [-0.8,2.8],
    [2.6,-1.6]
])

y_test = torch.tensor([0,1])

# PyTorch requires class labels to start with 0 so if we have 3 classes it would be 0,1 and 2
```

```python
# Now we will create a custom dataset class ToyDataset by subclassing from Pytorch's dataset parent class
from torch.utils.data import Dataset

class ToyDataset(Dataset):
    def __init__(self,X,y):
        self.features=X
        self.labels=y

    def __getitem__(self, index):   #1
        one_x=self.features[index]
        one_y=self.labels[index]
        return one_x,one_y

    def __len__(self):
        return self.labels.shape[0]  #2

train_ds = ToyDataset(X_train,y_train)
test_ds = ToyDataset(X_test,y_test)

#1 intructions to retrieve one data record and corresponding label
#2 instructions for returning the total length of the dataset.
```

```python
print(len(train_ds))
```

```python
from torch.utils.data import DataLoader

torch.manual_seed(123)

train_loader = DataLoader(
    dataset=train_ds, #1
    batch_size=2,
    shuffle=True,  #2
    num_workers=0  #3
)

test_loader = DataLoader(
    dataset=test_ds,
    batch_size=2,
    shuffle=False, #4
    num_workers=0
)

#1 the Toydataset instance created earlier serves as input to dta loader
#2 whether or not to shuffle the data
#3 number of background processes
#4 not necessary to shuffle t test dataset
```

```python
for idx, (x,y) in enumerate(train_loader):
    print(f" Batch {idx+1}:",x,y)
```

```python
# train_loader is iterating over training dataset and visiting each example only once in a batch of 2 hence last batch contains only one example.
# you can remove that by adding drop_last function

train_loader = DataLoader(
    dataset=train_ds, #1
    batch_size=2,
    shuffle=True,  #2
    num_workers=0,  #3
    drop_last=True
)

for idx, (x,y) in enumerate(train_loader):
    print(f" Batch {idx+1}:",x,y)
```

## A typical training loop

```python
import torch.nn.functional as F

torch.manual_seed(123)
model = NeuralNetwork(num_inputs=2, num_outputs=2)  #1
optimizer = torch.optim.SGD(model.parameters(),lr=0.5)  #2

num_epochs = 3

for epoch in range(num_epochs):

    model.train()
    for batch_idx,(features,labels) in enumerate(train_loader):
        logits = model(features)

        loss = F.cross_entropy(logits,labels)

        optimizer.zero_grad()  #3
        loss.backward()  #4
        optimizer.step()  #5

        ###LOGGING
        print(f"Epoch: {epoch+1:03d}/{num_epochs:03d}"
              f" | Batch {batch_idx:03d}/{len(train_loader):03d}"
              f" |Train Loss: {loss:.2f}")

    model.eval()

#1 the dataset has two features and two classes
#2 The optimizer needs to know which paramaetrs to optimize
#3 set the gradients from previous round to 0 to avoid unintended gradient accumulation
#4 Computes the gradient of the loss given the model parametrs
#5 the optimizer uses the gradients to update the model parametrs

```

```python
# Now that we have trained the model, lets use it for prediction
model.eval()
with torch.no_grad():
    outputs=model(X_train)

print(outputs)
```

```python
# to obtain class membership probabilities
torch.set_printoptions(sci_mode=False)
proba = torch.softmax(outputs,dim=1)
print(proba)
```

```python
# we can convert these values into class label using argmax, which returns the index position of highest value in each row if we set dim=1, dim=0 will return highest value for column
predictions = torch.argmax(proba,dim=1)
print(predictions)
```

```python
predictions = torch.argmax(outputs,dim=1)
print(predictions)
```

```python
def compute_accuracy(model,dataloader):
    model = model.eval()
    correct = 0.0
    total_examples = 0

    for idx,(features,labels) in enumerate(dataloader):

        with torch.no_grad():
            logits = model(features)

        predictions = torch.argmax(logits,dim=1)
        compare = labels==predictions
        correct+=torch.sum(compare)
        total_examples+=len(compare)

    return (correct/total_examples).item()
```

```python
print(compute_accuracy(model,train_loader))
```

```python
print(compute_accuracy(model,test_loader))
```

```python
import torch.nn.functional as F

torch.manual_seed(123)
model = NeuralNetwork(num_inputs=2, num_outputs=2)  

device = torch.device("cuda") #1
model = model.to("cuda") #2

optimizer = torch.optim.SGD(model.parameters(),lr=0.5)  

num_epochs = 3

for epoch in range(num_epochs):

    model.train()
    for batch_idx,(features,labels) in enumerate(train_loader):
        features,labels = features.to(device), labels.to(device)  #3
        logits = model(features)

        loss = F.cross_entropy(logits,labels)

        optimizer.zero_grad()  #3
        loss.backward()  #4
        optimizer.step()  #5

        ###LOGGING
        print(f"Epoch: {epoch+1:03d}/{num_epochs:03d}"
              f" | Batch {batch_idx:03d}/{len(train_loader):03d}"
              f" |Train Loss: {loss:.2f}")

    model.eval()

#1 define the variable device that defaults to GPU
#2 transfer the model ontp the GPU
#3 transfer the data to GPU
```
